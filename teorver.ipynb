# Задача 1.8 
В конкурсе по 5 номинациям участвуют 10 кино-фильмов. Сколько существует вариантов распределения призов, 
если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

```python

films = 10
nominations = 5


result_diff = 1
for _ in range(nominations):
    result_diff *= films

print('a)' , result_diff, "вариантов")

```

```python
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def combination(n, k):
    if k > n:
        return 0
    return factorial(n) // (factorial(k) * factorial(n - k))

result_alt = combination(films + nominations - 1, nominations)

print("б)", result_alt, "вариантов")
```

# Задача 1.9
Сколько существует семизначных чисел, состоя-щих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 — по 2 раза?
Реше н и е. Каждое семизначное число отличается от другого порядком следования цифр (причем и, - 3, п2 - 2, пз = 2, а их сумма равна 7), т.е. является перестановкой с повторениями из 7 эле-ментов

```python
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
total = 7
count4 = 3
count5 = 2
count6 = 2

answer = factorial(total) // (factorial(count4) * factorial(count5) * factorial(count6))

print(answer, "чисел")
```

# Задача 1.11
Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С

# Задача 1.38
Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

```python
import itertools
books = [1, 2, 3, 4, 5]
all_perms = list(itertools.permutations(books))

total = len(all_perms)
favorable = 1 

prob = favorable / total
print("Вероятность:", prob)
```

# Задача 1.43
Наудачу взятый телефонный номер состоит из 5 цифр.
Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не на-чинается с цифры ноль.

```python

total = 9 * (10 ** 4)

favorable_a = 9 * 9 * 8 * 7 * 6
prob_a = favorable_a / total
print("а) Различные:",  round(prob_a, 4))

favorable_b = 9
prob_b = favorable_b / total
print("б) Одинаковые:", prob_b)

favorable_c = 5 ** 5
prob_c = favorable_c / total
print("в) Нечетные:", round(prob_c, 5))
```

# Залание 1.48
В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выиг-рыша.

```python
total = 6 ** 3

count_11 = 0
for x in range(1, 7):
    for y in range(1, 7):
        for z in range(1, 7):
            if x + y + z == 11:
                count_11 += 1

prob_a = count_11 / total


count_win = 0
for x in range(1, 7):
    for y in range(1, 7):
        for z in range(1, 7):
            if x + y + z > 10:
                count_win += 1

prob_b = count_win / total

print("а) 11 очков:", prob_a)
print("б) Выигрыш (>10):", prob_b)
```
