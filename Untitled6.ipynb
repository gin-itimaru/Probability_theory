# Тема 1. Моделирование геометрической вероятности
Рассматривается бросок баскетбольного мяча в кольцо, закреплённое на щите. Мяч всегда попадает в щит (промах мимо щита исключён), причём все точки щита равновероятны. Требуется найти вероятность того, что мяч попадёт в кольцо.

# Теория
точка попадания мяча (x,y). Поскольку мяч всегда попадает в щит, пару (x,y) можно представить точкой на координатной плоскости.
Все возможные значения x и y образуют прямоугольник (щит):
0≤x≤18,0≤y≤10,5
Площадь щита:
    S щита =18⋅10,5=189



```python
import numpy as np
from matplotlib.patches import Rectangle, Circle
import matplotlib.pyplot as plt

def basketball(count):
    W, H = 18, 10.5          
    r = 2.25                 
    xc, yc = W/2, 0.15 + r   

    fig, ax = plt.subplots(figsize=(8, 5))
    ax.add_patch(Rectangle((0, 0), W, H, facecolor='#F5F5DC',
                            edgecolor='black', linewidth=2))          
    ax.add_patch(Rectangle((W/2 - 2.95, 0.15), 5.9, 4.5,
                            fill=False, edgecolor='red', linewidth=2)) 
    ax.add_patch(Circle((xc, yc), r, facecolor='orange', alpha=0.6,
                         edgecolor='darkorange', linewidth=2))         

    X = np.random.uniform(0, W, count)   
    Y = np.random.uniform(0, H, count)   

    inside = len([d for d in zip(X, Y)
                  if (d[0]-xc)**2 + (d[1]-yc)**2 <= r**2])

    colors = ['green' if (x-xc)**2 + (y-yc)**2 <= r**2 else 'red'
              for x, y in zip(X, Y)]
    ax.scatter(X, Y, s=6, c=colors, alpha=0.7)

    ax.set_xlim(-0.5, W + 0.5)
    ax.set_ylim(-0.5, H + 0.5)
    ax.set_aspect('equal')
    ax.set_title(f'Бросок мяча в кольцо: N = {count}\n'
                 f'Эмпирическая P = {inside/count:.4f}, '
                 f'теоретическая P = 0.0842', fontsize=12)
    ax.set_xlabel('Ширина щита (ед.)')
    ax.set_ylabel('Высота щита (ед.)')
    plt.grid(True, linestyle='--', alpha=0.4)
    plt.tight_layout()
    plt.show()

    print(f"Из {count} бросков благоприятными оказались {inside}. "
          f"Вероятность = {inside/count:.4f}")

basketball(1000)
basketball(10000)
basketball(100000)
```
