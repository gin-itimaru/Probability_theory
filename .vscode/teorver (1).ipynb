# Задача 1.8 
В конкурсе по 5 номинациям участвуют 10 кино-фильмов. Сколько существует вариантов распределения призов, 
если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

```python

films = 10
nominations = 5


result_diff = 1
for _ in range(nominations):
    result_diff *= films

print('a)' , result_diff, "вариантов")

```

```python
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def combination(n, k):
    if k > n:
        return 0
    return factorial(n) // (factorial(k) * factorial(n - k))

result_alt = combination(films + nominations - 1, nominations)

print("б)", result_alt, "вариантов")
```

# Задача 1.9
Сколько существует семизначных чисел, состоя-щих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 — по 2 раза?
Реше н и е. Каждое семизначное число отличается от другого порядком следования цифр (причем и, - 3, п2 - 2, пз = 2, а их сумма равна 7), т.е. является перестановкой с повторениями из 7 эле-ментов

```python
def factorial(n):
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result
total = 7
count4 = 3
count5 = 2
count6 = 2

answer = factorial(total) // (factorial(count4) * factorial(count5) * factorial(count6))

print(answer, "чисел")
```

# Задача 1.11
Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С

```python
import math

total = math.factorial(6) // (math.factorial(3) * math.factorial(2) * math.factorial(1))
prob = 1 / total

print("Вероятность слова 'АНАНАС':", prob)
```

# Задача 1.16

# Задача 1.38
Пятитомное собрание сочинений расположено на полке в случайном порядке. Какова вероятность того, что книги стоят слева направо в порядке нумерации томов (от 1 до 5)?

```python
import itertools
books = [1, 2, 3, 4, 5]
all_perms = list(itertools.permutations(books))

total = len(all_perms)
favorable = 1 

prob = favorable / total
print("Вероятность:", prob)
```

# Задача 1.43
Наудачу взятый телефонный номер состоит из 5 цифр.
Какова вероятность того, что в нем все цифры: а) различные; б) одинаковые; в) нечетные? Известно, что номер телефона не на-чинается с цифры ноль.

```python

total = 9 * (10 ** 4)

favorable_a = 9 * 9 * 8 * 7 * 6
prob_a = favorable_a / total
print("а) Различные:",  round(prob_a, 4))

favorable_b = 9
prob_b = favorable_b / total
print("б) Одинаковые:", prob_b)

favorable_c = 5 ** 5
prob_c = favorable_c / total
print("в) Нечетные:", round(prob_c, 5))
```

# Залание 1.48
В старинной игре в кости необходимо было для выигрыша получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выиг-рыша.

```python
total = 6 ** 3

count_11 = 0
for x in range(1, 7):
    for y in range(1, 7):
        for z in range(1, 7):
            if x + y + z == 11:
                count_11 += 1

prob_a = count_11 / total


count_win = 0
for x in range(1, 7):
    for y in range(1, 7):
        for z in range(1, 7):
            if x + y + z > 10:
                count_win += 1

prob_b = count_win / total

print("а) 11 очков:", prob_a)
print("б) Выигрыш (>10):", prob_b)
```

# Задача про шарики 
Решить классически и смоделировать решение задачи.

В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.

# Классическое решение
Рассмотрим событие 
 - шар, вытянутый из второй урны - белый

```python
from math import comb

u1_white, u1_black = 15, 5
u2_white, u2_black = 6, 1
k = 3

n1 = u1_white + u1_black
n2_new = u2_white + u2_black + k
total_ways = comb(n1, k)

ans = 0.0

for w_moved in range(k + 1):
    b_moved = k - w_moved
    p_move = (comb(u1_white, w_moved) * comb(u1_black, b_moved)) / total_ways
    p_draw = (u2_white + w_moved) / n2_new
    ans += p_move * p_draw

ans
```

# Моделирование решения задачи

```python
%mamba install pandas
import random
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
ans = 9405 / 11400


def run_simulation(trials):
    urn1 = [1] * 15 + [0] * 5
    success = 0
    for _ in range(trials):
        moved = random.sample(urn1, 3)
        w2_current = 6 + sum(moved)
        if random.random() < (w2_current / 10.0):
            success += 1
    return success / trials


trial_counts = [10, 50, 100, 500, 1000, 5000, 10000, 50000, 100000, 1000000]
data = []

for t in trial_counts:
    sim_p = run_simulation(t)
    deviation = abs(sim_p - ans)
    data.append({"Испытания": t, "Моделирование": sim_p, "Отклонение": deviation})

df = pd.DataFrame(data)
print(df.to_string(index=False))

plt.figure(figsize=(10, 6))
plt.plot(df["Испытания"], df["Моделирование"], marker='o', linestyle='-', color='b', label='Эксперимент')
plt.axhline(y=ans, color='r', linestyle='--', label=f'Теория ({ans:.4f})')

plt.xscale('log')
plt.xlabel('Число испытаний (N)')
plt.ylabel('Вероятность белого шара')
plt.title('Сходимость частоты эксперимента к истинной вероятности')
plt.grid(True, which="both", linestyle="--", alpha=0.5)
plt.legend()
plt.show()
```
